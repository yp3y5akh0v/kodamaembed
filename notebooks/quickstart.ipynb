# KodamaEmbed quickstart

Embed passages, search them and rerank the results.

```python
%pip install kodamaembed
```

```python
import kodamaembed

embedder = kodamaembed.Embedder("sentence-transformers/all-MiniLM-L6-v2", device="auto")
passages = [
    "Berlin is the capital of Germany.",
    "The Eiffel Tower is in Paris.",
    "Bananas are rich in potassium.",
    "Solar panels turn sunlight into electricity.",
    "The Pacific is the largest ocean on Earth.",
]
vectors = embedder.embed_passages(passages)
question = "Which city is the capital of Germany?"
scores = vectors @ embedder.embed_queries(question)[0]
for index in scores.argsort()[::-1]:
    print(f"{scores[index]:6.3f}  {passages[index]}")
```

```python
reranker = kodamaembed.Reranker("cross-encoder/ms-marco-MiniLM-L6-v2", device="auto")
for index, score in reranker.rerank(question, passages, top_k=3):
    print(f"{score:7.3f}  {passages[index]}")
```

`device="auto"` takes CUDA at compute capability 8.0 or newer, else a discrete Vulkan GPU,
else the processor.

```python
print(kodamaembed.__version__, embedder.device, reranker.device)
```

A text's vector does not depend on its batch, and `fingerprint` identifies the model files,
input handling and numerics behind it.

```python
alone = embedder.embed_passages(passages[1])
assert alone[0].tobytes() == vectors[1].tobytes()
print(embedder.fingerprint)
```
